Wprowadzenie miesiąca jako zmiennej kategorycznej

```python
X['Date'] = pd.to_datetime(X['Date'],format='%m/%d/%Y')
X['Month'] = X['Date'].dt.month.astype('category')
```

Przygotowanie targetu:

```python
from sklearn.impute import SimpleImputer

gt_columns = ['CO(GT)','C6H6(GT)','NOx(GT)','NO2(GT)']
imputer = SimpleImputer(strategy='median')
X[gt_columns] = imputer.fit_transform(X[gt_columns])

y = X[gt_columns].sum(axis=1)
X = X.drop(columns=gt_columns+ ['Date'])
```

Pozostały preprocessing danych - one hot encoding na zmiennej Month oraz imputacja dla danych z X

```python
from sklearn.preprocessing import OneHotEncoder

cols_to_impute = X.columns.drop('Month')
imputer_x = SimpleImputer(strategy='median')
X[cols_to_impute] = imputer_x.fit_transform(X[cols_to_impute])

month_encoder = OneHotEncoder(drop='first',sparse_output=False)
month_encoded = month_encoder.fit_transform(X[['Month']])

# needed to match the indices in joining X (because of removal rows in earlier in this dataset)
month_encoded_df = pd.DataFrame(
    month_encoded,
    columns=month_encoder.get_feature_names_out(['Month']),
    index=X.index
)

X_ohe = X.drop(columns=['Month']).join(month_encoded_df)
```

Usuwamy jedną z pary silnie skorelowanych zmiennych

```python
def remove_highly_correlated_features(df, threshold=0.8):
  corr_matrix = df.corr()
  upper_tri = corr_matrix.where(np.triu(np.ones(corr_matrix.shape),k=1).astype(bool))
  to_drop = {
      column if df[column].var() < df[feature].var() else feature
      for column in upper_tri.columns
      for feature in upper_tri.index[upper_tri[column] >= threshold]
  }
  return df.drop(columns=to_drop)
```

```python
X_uncorr = remove_highly_correlated_features(X_ohe,0.8)
```

## Metody liniowe

funkcja do ewaluacji modeli

```python
import time
from sklearn.metrics import mean_squared_error, mean_absolute_error

def evaluate_model(model, X_train, y_train, X_test, y_test):
  start_time = time.time()
  model.fit(X_train, y_train)
  end_time = time.time()
  train_time = end_time - start_time

  y_pred_train = model.predict(X_train)
  y_pred_test = model.predict(X_test)

  rmse_train = mean_squared_error(y_train, y_pred_train)
  rmse_test = mean_squared_error(y_test, y_pred_test)
  mae_train = mean_absolute_error(y_train, y_pred_train)
  mae_test = mean_absolute_error(y_test, y_pred_test)

  print(f"Training Time: {train_time:.4f}")
  print(f"Train RMSE: {rmse_train:.2f}")
  print(f"Test RMSE: {rmse_test:.2f}")
  print(f"Train MAE: {mae_train:.2f}")
  print(f"Test MAE: {mae_test:.2f}")
```

```python
def grid_search_model(model, param_grid, X_train, y_train, X_test, y_test):
  grid_search = GridSearchCV(model, param_grid, scoring='neg_mean_squared_error',cv=5,n_jobs=-1)
  start_time = time.time()
  grid_search.fit(X_train, y_train)
  end_time = time.time()
  train_time = end_time - start_time

  best_model = grid_search.best_estimator_
  best_params = grid_search.best_params_
  print(f"Training Time: {train_time:.4f}")
  print(f"Best Parameters: {best_params}")

  return best_model
```

Regresja liniowa

```python
from sklearn.linear_model import LinearRegression

lr_model = LinearRegression()
evaluate_model(lr_model, X_train, y_train, X_test, y_test)
```

model z regularyzacją L1

```python
from sklearn.linear_model import Lasso

lasso_model = Lasso()
print(f"alpha: {lasso_model.alpha}")
evaluate_model(lasso_model, X_train, y_train, X_test, y_test)
```

```python
from sklearn.model_selection import GridSearchCV

lasso_grid = {'alpha':[0.001, 0.01, 0.1, 1.0, 10.0, 100.0]}

lasso_best_model = grid_search_model(Lasso(), lasso_grid, X_train, y_train, X_test, y_test)
evaluate_model(lasso_best_model, X_train, y_train, X_test, y_test)
```

model z regularyzacją l2

```python
from sklearn.linear_model import Ridge

ridge_model = Ridge()
print(f"alpha: {ridge_model.alpha}")
evaluate_model(ridge_model,X_train, y_train, X_test, y_test)
```

```python
ridge_grid = {'alpha':[0.001, 0.01, 0.1, 1.0, 10.0, 100.0]}

ridge_best_model = grid_search_model(Ridge(), ridge_grid, X_train, y_train, X_test, y_test)
evaluate_model(ridge_best_model, X_train, y_train, X_test, y_test)
```

regresja Hubera

```python
from sklearn.linear_model import HuberRegressor

huber_model = HuberRegressor()
print(f"epsilon: {huber_model.epsilon}, alpha: {huber_model.alpha}")
evaluate_model(huber_model, X_train, y_train, X_test, y_test)
```

```python
huber_grid = {
    'epsilon': [1.20, 1.35, 1.5, 1.65],
    'alpha': [0.0001, 0.001, 0.01, 0.1, 1.0]
}

huber_best_model = grid_search_model(HuberRegressor(), huber_grid, X_train, y_train, X_test, y_test)
evaluate_model(huber_best_model, X_train, y_train, X_test, y_test)
```

Dla każdego z modeli można zauważyć, że wartości RMSE oraz MAE znacznie się od siebie różnią. RMSE podnosi do kwadratu błędy przed ich uśrenieniem, co tłumaczy dlaczego te dwie metryki mogą się tak znacząco różnić .Wskazuje to na obecność w zbiorze wartości znacznie odstających wartościami od średniej, z czym modele sobie źle poradziły.

Czasy treningu oraz przeszukiwania grid search były porównywalne między regresją liniową, modelem z regularyzacją L1 oraz regularyzajcą L2, natomiast dla regresji Hubera czas ten był znacznie dłuższy, dlatego, że model ten jest bardziej skomplikowany oraz wymaga bardziej złożonej optymalizacji.

Wyniki regresji liniowej, modelu z regularyzacją L1 oraz L2 są do siebie zbliżone, jeśli chodzi o MAE to utrzymują się w przedziale 90.5-91, dla RMSE osiągają ok 14760. Odbiegają od nich wyniki regresji Hubera, które są niższe dla MAE oraz wyższe dla RMSE, odpowiednio 88.22 oraz 14963 dla modelu przed optymalizajcą, 88.70 i 14790 dla modelu zoptymalizowanego. Wynika to z własności regresji Hubera, która jest bardziej odporna na odstające wartości, niż pozostałe metody, dzięki czemu lepiej przewiduje typowe wartości - co odzwierciedla niższe MAE, ale gorzej radzi sobie z ekstremalnymi wartościami - wyższe RMSE.

## Badanie wpływu wstępnego przetwarzania danych na wyniki

użyję regresji Hubera, ponieważ dawała najlepsze wyniki w poprzedniej części zadania

```python
from sklearn.pipeline import Pipeline

X_train_full, X_test_full, y_train_full, y_test_full = train_test_split(X, y, test_size=0.2, random_state=0)
```

```python
pipe_1 = Pipeline([('model',HuberRegressor(epsilon=1.65, alpha=0.0001, max_iter=2000))])
pipe_1.fit(X_train_full, y_train_full)
evaluate_model(pipe_1, X_train_full, y_train_full, X_test_full, y_test_full)
```

```python
pipe_2 = Pipeline([
    ('scaler',StandardScaler()),('model',HuberRegressor(epsilon=1.65, alpha=0.0001, max_iter=2000))])
pipe_2.fit(X_train_full, y_train_full)
evaluate_model(pipe_2, X_train_full, y_train_full, X_test_full, y_test_full)
```

```python
X_train_uncorr, X_test_uncorr, y_train_uncorr, y_test_uncorr = train_test_split(X_uncorr, y, test_size=0.2, random_state=0)
pipe_3 = Pipeline([('model',HuberRegressor(epsilon=1.65, alpha=0.0001, max_iter=2000))])
pipe_3.fit(X_train_uncorr, y_train_uncorr)
evaluate_model(pipe_3, X_train_uncorr, y_train_uncorr, X_test_uncorr, y_test_uncorr)
```

```python
X_train_ohe, X_test_ohe, y_train_ohe, y_test_ohe = train_test_split(X_ohe, y, test_size=0.2, random_state=0)
pipe_4 = Pipeline([('model',HuberRegressor(epsilon=1.65, alpha=0.0001, max_iter=2000))])
pipe_4.fit(X_train_ohe, y_train_ohe)
evaluate_model(pipe_4, X_train_ohe, y_train_ohe, X_test_ohe, y_test_ohe)
```

```python
pipe_5 = Pipeline([('scaler',StandardScaler()),('model',HuberRegressor(epsilon=1.65, alpha=0.0001, max_iter=2000))])
pipe_5.fit(X_train_ohe, y_train_ohe)
evaluate_model(pipe_5, X_train_ohe, y_train_ohe, X_test_ohe, y_test_ohe)
```

Po analizie powyższych wyników, można dokonać ciekawej obserwacji: usunięcie skorelowanych cech (na poziomie 0.8 bądź wyższym), znacznie pogorszyło wyniki regresji hubera. Wszystkie modele które korzystały z danych bez usuniętych cech skorelowanych uzyskały znacznie lepsze wyniki, a najlepiej porazdził sobie model znormalizowany oraz z zastosowanym one hot encodingiem dla zmiennej kateogrycznej (miesiąca). Można z tego wyciągnąć wniosek, że został dobrany zbyt niski threshold, który posłużył do usuwania skorelowanych cech. Usunięte kolumny zawierały dane, które mimo korelacji z innymi, zawierały wartościowe dane pozwalające modelowi uzyskać lepsze wyniki.

## Wpływ wartości parametrów na osiągi modelu

```python
coef_lr = lr_model.coef_
coef_lasso = lasso_best_model.coef_
coef_ridge = ridge_best_model.coef_
coef_huber = huber_best_model.coef_

df_coefs = pd.DataFrame({
    'Linear Regression': coef_lr,
    'Lasso (L1)': coef_lasso,
    'Ridge (L2)': coef_ridge,
    'Huber Regressor': coef_huber
}, index=X_train.columns)

ax = df_coefs.plot(kind='bar', figsize=(15, 6), colormap='Set2', width=0.8)
plt.title('Porównanie wartości współczynników dla poszczególnych modeli', fontsize=15)
plt.ylabel('Wartość współczynnika', fontsize=12)
plt.xlabel('Cechy', fontsize=12)
plt.axhline(0, color='black', linewidth=1)
plt.legend(loc='best')
plt.grid(axis='y', alpha=0.7)
plt.tight_layout()
plt.show()
```

Można zauważyć, wszystkie modele poza regresją hubera mają prawie identyczne współczynniki, co tłumaczy ich bardzo zbliżone wyniki RMSE oraz MAE. Regresja hubera różni się zauważalnie, co również odzwierciedla się w wynikach modelu

```python
coefs_pipe = pipe_5.named_steps['model'].coef_

df_coefs = pd.DataFrame({
    'Model z Pipeline': coefs_pipe
}, index=X_train_ohe.columns)

ax = df_coefs.plot(kind='bar', figsize=(15, 6), colormap='Set2', width=0.8)
plt.title('Wartości współczynników dla modelu z Pipeline', fontsize=15)
plt.ylabel('Wartość współczynnika', fontsize=12)
plt.xlabel('Cechy', fontsize=12)
plt.axhline(0, color='black', linewidth=1)
plt.legend(loc='best')
plt.grid(axis='y', alpha=0.7)
plt.tight_layout()
plt.show()
```
